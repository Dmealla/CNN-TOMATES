Tiene que dar 0.247. Repaso rápido de qué hace: multiplica cada color por su peso, suma todo y, si el resultado es negativo, lo deja en 0. Con estos pesos se activa cuando hay más verde que rojo y azul.

```python
import numpy as np
#pixel del centro de hoja sana: rojo, verde,azul (llevados a escala de 0 a 1)
entradas= np.array([132,158,121])/255

#el verde suma, el rojo y azul restan
pesos= np.array([2.0, -1.0,-1.0])

suma= (entradas*pesos).sum()
salida= max(0, suma) #ReLU si es negativo, se pone a cero
print(salida)
```

Fijate en el 0 de la hoja sana. La suma dio negativa (−0,059), porque el verde y el azul juntos superan al doble del rojo, y la regla ReLU la convirtió en 0. Así se ve una neurona “apagada”: cuando lo que busca no está, no dice nada.

Y mirá lo que lograste: con solo cambiar los pesos, la misma neurona pasó a detectar otra cosa. Los pesos son lo que la neurona sabe. Entrenar una red es encontrar los pesos adecuados para miles de neuronas a la vez.
