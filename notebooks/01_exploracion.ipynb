contar las fotos de cada clase
Antes de entrenar hay que saber cuántos ejemplos hay de cada clase. En celdas nuevas del mismo notebook:

```python
import pandas as pd
filas =[]
for carpeta in DATOS.iterdir():
    if carpeta.name.startswith("Tomato"):
        cantidad= len (list(carpeta.iterdir()))
        filas.append({"clase":carpeta.name, "cantidad":cantidad})
conteo = pd.DataFrame(filas).sort_values("cantidad", ascending=False)
conteo        
```

```python
conteo.plot.barh(x="clase", y="cantidad", legend=False, title="Fotos por clase de (tomate)")
```
